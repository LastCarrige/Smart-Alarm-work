```python
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d
import os
```

```python
USER_IDS = [46343, 759667, 781756, 844359, 1066528, 1360686, 1449548, 1455390, 1818471, 2598705, 2638030, 3509524, 3997827, 4018081, 4314139, 4426783, 5132496, 5383425, 5498603, 5797046, 6220552, 7749105, 8000685, 8173033, 8258170, 8530312, 8686948, 8692923, 9106476, 9618981, 9961348]
```

```python
INPUT_DIR = "../data/movement/"
OUTPUT_DIR = "../data/processed/"
os.makedirs(OUTPUT_DIR, exist_ok=True)

for uid in USER_IDS:
    print(f"[{uid}] Обробка акселерометра...")
    try:
        file_path = f"{INPUT_DIR}{uid}_acceleration.txt"
        if not os.path.exists(file_path):
            print(f"  -> Пропуск: файл {file_path} не знайдено.")
            continue

        df = pd.read_csv(file_path, sep=r'\s+', header=None, names=['time', 'x', 'y', 'z'])
        df['enmo'] = np.maximum(0, np.sqrt(df['x']**2 + df['y']**2 + df['z']**2) - 1.0)

        df_clean_motion = df[['time', 'enmo']]
        out_path = f"{OUTPUT_DIR}{uid}_clean_motion.csv"
        df_clean_motion.to_csv(out_path, index=False)
        print(f"  -> Збережено: {out_path}")

    except Exception as e:
        print(f"  -> ПОМИЛКА у юзера {uid}: {e}")
```
