```python
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d
import os
```

```python
USER_IDS = [46343, 759667, 781756, 844359, 1066528, 1360686, 1449548, 1455390, 1818471, 2598705, 2638030, 3509524, 3997827, 4018081, 4314139, 4426783, 5132496, 5383425, 5498603, 5797046, 6220552, 7749105, 8000685, 8173033, 8258170, 8530312, 8686948, 8692923, 9106476, 9618981, 9961348]
```

```python
INPUT_DIR = "../data/heartrate/"
OUTPUT_DIR = "../data/processed/"
os.makedirs(OUTPUT_DIR, exist_ok=True)

for uid in USER_IDS:
    print(f"[{uid}] Обробка пульсу...")
    try:
        file_path = f"{INPUT_DIR}{uid}_heartrate.txt"
        if not os.path.exists(file_path):
            print(f"  -> Пропуск: файл {file_path} не знайдено.")
            continue

        df = pd.read_csv(file_path, header=None, names=['time', 'hr'])
        df_processed = df.groupby('time')['hr'].mean().reset_index()


        df_processed['hr'] = np.clip(df_processed['hr'], 35, 170)

        df_processed['rr'] = 60000 / df_processed['hr']

        df_clean_hr = df_processed[['time', 'rr']]

        '''
        time = df_processed['time'].values
        hr = df_processed['hr'].values

        regular_time = np.arange(time.min(), time.max(), 1.0)
        spline_func = interp1d(time, hr, kind='cubic', fill_value="extrapolate")
        regular_hr = spline_func(regular_time)

        regular_hr = np.clip(regular_hr, 35, 170)
        regular_rr = 60000 / regular_hr

        df_clean_hr = pd.DataFrame({'time': regular_time, 'rr': regular_rr})
        '''

        out_path = f"{OUTPUT_DIR}{uid}_clean_hr.csv"
        df_clean_hr.to_csv(out_path, index=False)
        print(f"  -> Збережено: {out_path}")

    except Exception as e:
        print(f"  -> ПОМИЛКА у юзера {uid}: {e}")
```
